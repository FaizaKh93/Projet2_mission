# Conclusion de l'exploration

Les cinq fichiers EdStats ont été profilés selon la même grille : granularité, volumétrie,
doublons, valeurs manquantes, statistiques descriptives et variables catégorielles.

| Fichier | Lignes | Colonnes | Dont numériques | Doublons | Une ligne représente |
|---|---:|---:|---:|---:|---|
| `EdStatsCountry` | 241 | 32 | 3 | 0 | un pays, ou un agrégat géographique / économique |
| `EdStatsCountry-Series` | 613 | 4 | 0 | 0 | un couple pays × indicateur et sa source |
| `EdStatsSeries` | 3 665 | 21 | 0 | 0 | un indicateur, sa définition et sa catégorie |
| `EdStatsFootNote` | 643 638 | 5 | 0 | 0 | une note sur un triplet pays × indicateur × année |
| `EdStatsData` | 886 930 | 70 | 65 | 0 | un couple pays × indicateur, les colonnes portant les années |

**Aucun doublon dans aucun des cinq fichiers.** De ce point de vue, le jeu de données est sain.

**Trois fichiers sur cinq ne contiennent aucune colonne numérique.** `EdStatsCountry-Series`,
`EdStatsSeries` et `EdStatsFootNote` sont purement documentaires : ils décrivent et annotent,
ils ne mesurent pas. Toute la matière chiffrée se trouve dans `EdStatsData`, dont 65 des
70 colonnes sont des années.

**`EdStatsCountry` ne contient pas que des pays.** 27 de ses 241 lignes ne respectent pas le
critère retenu — un code alpha-2 de deux lettres valide et une région renseignée — et toutes
ont en commun une région vide. Vingt-cinq sont des agrégats (`WLD` World, `SSF` Sub-Saharan
Africa, `LIC` Low income…) ; les deux dernières, Gibraltar et Nauru, sont de vrais
territoires que la table laisse sans région. Conserver les agrégats fausserait toute
comparaison entre pays, puisqu'un agrégat cumule les valeurs de dizaines d'entre eux.
Il reste **214 pays** exploitables.

Les colonnes entièrement vides ont par ailleurs été écartées de chaque fichier.

## Ce que cette exploration a produit

La grille d'analyse s'est révélée identique pour les cinq fichiers. C'est ce constat qui a
permis de la factoriser dans la fonction `DataProcessing` de `src/fonctions_exercice1.py`,
appliquée ensuite aux cinq fichiers en une ligne chacun. La liste des 27 faux pays, elle,
sert de filtre à toutes les tables en aval.
