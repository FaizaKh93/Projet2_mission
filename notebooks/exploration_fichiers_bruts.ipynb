## Etape 2 
## 2.1 Ce que représente chaque ligne : 
Chaque ligne représente un pays avec ces informations : code, nom court, etc.
## 2.2 Nombre de lignes et de colonnes

```python
EdStatsCountry.shape
```

```python
print("le nombre de lignes est de", EdStatsCountry.shape[0], "et le nombre de colonnes est de", EdStatsCountry.shape[1])
```

## 2.3 Nombre de doublons dans le jeu de données

```python
Nb_doublons_Country = EdStatsCountry.duplicated().sum()
print("le nombre de doublons est de",Nb_doublons_Country)
```

### Suppression des doublons (inutile dans ce cas, mais à garder pour les autres fichiers csv)

```python
EdStatsCountry = EdStatsCountry.drop_duplicates()
EdStatsCountry.shape
#EdStatsSeries.head()
```

## 2.4 Calcul de la proportion de valeurs manquantes par colonne

```python
valeursManquantes_Country = EdStatsCountry.isna()
valeursManquantes_Country.head()
```

```python
valeursManquantes_moyenne_Country = EdStatsCountry.isna().mean()*100
valeursManquantes_moyenne_Country
```

### suppression des colonnes vides où valeursManquantes_moyenne = 100%

```python
EdStatsSeriesCountry_clean = EdStatsCountry.drop(columns=valeursManquantes_moyenne_Country[valeursManquantes_moyenne_Country==100].index)
EdStatsSeriesCountry_clean
```

## 2.5 Calcul des statistiques descriptives basiques en utilisant describe() pour les colonnes numériques 

```python
EdStatsSeriesCountry_clean.describe()
```

```python
#colonnes_numeriques = EdStatsSeries_clean[EdStatsSeries_clean.select_dtypes(include = 'number')]
#colonnes_numeriques.sample(5)
#colonnes_numeriques_Country = EdStatsSeriesCountry_clean[EdStatsSeriesCountry_clean.select_dtypes(include = 'number').columns.tolist()]
colonnes_numeriques_Country = EdStatsSeriesCountry_clean.select_dtypes(include = 'number')

print("nombre de colonnes numérique :",colonnes_numeriques_Country.shape[1])
```

```python
#EdStatsSeriesCountry_clean.info(verbose=True)
colonnes_numeriques_Country.info(verbose=True)
```

## 2.6 Calcul du nombre d'occurrences de chaque valeur possible de la colonne pour les colonnes catégorielles 

```python
colonnes_categorielles_Country = EdStatsSeriesCountry_clean.select_dtypes(include = 'object')
colonnes_categorielles_Country.shape
```

```python
for col_cat in colonnes_categorielles_Country:
    print(EdStatsSeriesCountry_clean[col_cat].value_counts())
    print("==============================================================================================")
    print("==============================================================================================")
```
