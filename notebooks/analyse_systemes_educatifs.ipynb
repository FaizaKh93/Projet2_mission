### 3.2 Agrégez ce dataframe pour en construire un nouveau : chaque ligne doit correspondre à un pays et chaque colonne doit correspondre à un indicateur.
Le choix s'est porté sur la prise en compte de la dernière valeur renseignée car cela reflète la situation actuelle du pays, certains indicateurs évolues beaucoup à travers les années.

```python
#from fonctions_exercice1 import derniere_valeur_non_nulle
def derniere_valeur_non_nulle(row):
    # On récupère les valeurs non NaN et non nulles
    valeurs = row.dropna() # supprimer les valeurs manquantes
    
    #derniere_valeur = derniere_valeur[derniere_valeur != 0] # filtrer les valeurs nulles

    # s'il n'y a plus de valeurs
    if valeurs.empty:
        return np.nan

    # s'il y a une seul valeur récupérée
    if len(valeurs) ==1:
        return valeurs.iloc[0]

    # S'il y a plusieurs valeurs ==> filtrage des outliers
    Q1 = valeurs.quantile(0.25)
    Q3 = valeurs.quantile(0.75)
    IQR = Q3 - Q1

    borne_basse = Q1 - 1.5 * IQR
    borne_haute = Q3 + 1.5 * IQR
        
    # filtrage des outliers
    valeurs_filtrees = valeurs[(valeurs >= borne_basse) & (valeurs <= borne_haute)]
    
    # s'il n'y a plus de valeurs
    if  valeurs_filtrees.empty:
        valeurs_filtrees =valeurs  

    # prendre la dernière valeur des valeurs filtrées
    return valeurs_filtrees.iloc[-1]

#annees_filtre = EdStastDataCountry_filtre_indicateurs_finaux.columns[4:]
#display(annees_filtre)
annees_filtre = [col for col in EdStastDataCountry_filtre_indicateurs_finaux.columns if col.isdigit() and 1999 <= int(col) <= 2015]

print("Colonnes d'années retenues :", annees_filtre)

```

```python
EdStastDataCountry_filtre_indicateurs_finaux["Valeur Finale"] =EdStastDataCountry_filtre_indicateurs_finaux[annees_filtre].apply(derniere_valeur_non_nulle, axis=1)
EdStastDataCountry_filtre_indicateurs_finaux.head()
```

```python
# création du dataframe final
EdStatsDataCountryIndicator = EdStastDataCountry_filtre_indicateurs_finaux.pivot_table(index = "Country Name", columns = "Indicator Name", values = "Valeur Finale")
EdStatsDataCountryIndicator.head()
```

### 2.1.3 Proposez une méthode quantitative, sur la base des indicateurs, pour sélectionner des pays et répondre à la demande de Mark  
D'après l'analyse de la distribution des indicateurs choisis, un pays candidat présenterait des valeurs entre Q1 et Q3, pas très dévelopé, car risque de saturation et pas de pays à faible revenu, car marché instable. 
Les scores de chaque indicateur sont calculés en écartant les pays dévlopé et les pays à faible revenu (score 0), car ils ne sont pas ciblés.

```python
# fonction pour calculer le score
def score_indicateur(data_indicateur):
    Q1 = data_indicateur.quantile(0.25)
    Q3 = data_indicateur.quantile(0.75)

    # choix de la zone à éviter
    masque_score = (data_indicateur<Q1) | (data_indicateur>Q3)
    
    score_indic = (data_indicateur - Q1) / (Q3 - Q1)   # normatlisation dans cette zone (entre Q1 et Q3)

    score_indic[masque_score] = 0     

    return score_indic
```

```python
# calcul du score des indicateurs pour chaque pays
Data_scores_indicateurs = EdStatsDataCountryIndicator_filtre.apply(score_indicateur)
Data_scores_indicateurs = Data_scores_indicateurs.fillna(0)
display(Data_scores_indicateurs)
```

### Méthode de classement de score avec calcul simple de la moyenne

```python
Data_scores_indicateurs["Score Final"] = Data_scores_indicateurs.mean(axis=1)
classement_moyenne = Data_scores_indicateurs["Score Final"].sort_values(ascending=False)  # pas très représentatif, car en réalité les poids des indicateurs ne sont pas les mêmes
display(classement_moyenne)
```
